```python
import pandas as pd 
cust_df = pd.read_csv('dataset/olist_customers_dataset.csv')
loc_df = pd.read_csv('dataset/olist_geolocation_dataset.csv')
item_df = pd.read_csv('dataset/olist_order_items_dataset.csv')
pay_df = pd.read_csv('dataset/olist_order_payments_dataset.csv')
review_df = pd.read_csv('dataset/olist_order_reviews_dataset.csv')
order_df = pd.read_csv('dataset/olist_orders_dataset.csv')
prod_df = pd.read_csv('dataset/olist_products_dataset.csv')
sell_df = pd.read_csv('dataset/olist_sellers_dataset.csv')
trans_df = pd.read_csv('dataset/product_category_name_translation.csv')
```

```python
cust_df['customer_state'] = cust_df['customer_state'].map(cust_df.groupby('customer_state')['customer_zip_code_prefix'].mean())
cust_df['customer_city'] = cust_df['customer_city'].map(cust_df.groupby('customer_city')['customer_zip_code_prefix'].mean())
```

```python
loc_df = loc_df[loc_df['geolocation_zip_code_prefix'].duplicated() == False]
```

```python
loc_df['geolocation_state'] = loc_df['geolocation_state'].map(loc_df.groupby('geolocation_state')['geolocation_zip_code_prefix'].mean())
loc_df['geolocation_city'] = loc_df['geolocation_city'].map(loc_df.groupby('geolocation_city')['geolocation_zip_code_prefix'].mean())
```

```python
item_df['shipping_limit_date'] = pd.to_datetime(item_df['shipping_limit_date'])
```

```python
pay_df = pay_df[(pay_df['payment_value'] > 0) & (pay_df['payment_installments'] > 0)]
```

```python
pay_df = pd.get_dummies(pay_df, columns=['payment_type'], dtype=int)
```

```python
pay_df.head()
```

```python
review_df = review_df[review_df['review_id'].duplicated() == False]
```

```python
review_df.drop(columns=['review_comment_title', 'review_comment_message'], inplace=True)
```

```python
order_df['order_purchase_timestamp'] = pd.to_datetime(order_df['order_purchase_timestamp'])
order_df['order_approved_at'] = pd.to_datetime(order_df['order_approved_at'])
order_df['order_delivered_carrier_date'] = pd.to_datetime(order_df['order_delivered_carrier_date'])
order_df['order_delivered_customer_date'] = pd.to_datetime(order_df['order_delivered_customer_date'])
order_df['order_estimated_delivery_date'] = pd.to_datetime(order_df['order_estimated_delivery_date'])
```

```python
order_df = pd.get_dummies(order_df, columns=['order_status'], dtype=int)
```

```python
order_df.dropna(inplace=True)
```

```python
prod_df.dropna(inplace=True)
```

```python
prod_df['product_photos_qty'] = prod_df['product_photos_qty'].astype(int)
prod_df['product_name_lenght'] = prod_df['product_name_lenght'].astype(int)
prod_df['product_description_lenght'] = prod_df['product_description_lenght'].astype(int)
```

```python
prod_df = prod_df[prod_df['product_weight_g'] > 0]
```

```python
sell_df['seller_state'] = sell_df['seller_state'].map(sell_df.groupby('seller_state')['seller_zip_code_prefix'].mean())
sell_df['seller_city'] = sell_df['seller_city'].map(sell_df.groupby('seller_city')['seller_zip_code_prefix'].mean())
```

```python
cust_df.to_csv('cleaned_dataset/olist_customers_dataset.csv', index=False)
loc_df.to_csv('cleaned_dataset/olist_geolocation_dataset.csv', index=False)
item_df.to_csv('cleaned_dataset/olist_order_items_dataset.csv', index=False)
pay_df.to_csv('cleaned_dataset/olist_order_payments_dataset.csv', index=False)
review_df.to_csv('cleaned_dataset/olist_order_reviews_dataset.csv', index=False)
order_df.to_csv('cleaned_dataset/olist_orders_dataset.csv', index=False)
prod_df.to_csv('cleaned_dataset/olist_products_dataset.csv', index=False)
sell_df.to_csv('cleaned_dataset/olist_sellers_dataset.csv', index=False)
trans_df.to_csv('cleaned_dataset/product_category_name_translation.csv', index=False)
```
