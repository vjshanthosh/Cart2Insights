```python
for i in create_tables:
    cursor.execute(create_tables[i]['create_sql'])
    cursor.executemany(create_tables[i]['insert_sql'], [tuple(j) for j in pd.read_csv(i).values])
```

```python
cursor.execute("DELETE c FROM customers AS c LEFT JOIN locations AS l ON c.customer_zip_code_prefix = l.geolocation_zip_code_prefix " \
"WHERE l.geolocation_zip_code_prefix IS NULL")
cursor.execute("""ALTER TABLE customers ADD FOREIGN KEY(customer_zip_code_prefix) REFERENCES locations(geolocation_zip_code_prefix)""")
```

```python
cursor.execute("DELETE s FROM sellers AS s LEFT JOIN locations AS l ON s.seller_zip_code_prefix = l.geolocation_zip_code_prefix " \
"WHERE l.geolocation_zip_code_prefix IS NULL")
cursor.execute("""ALTER TABLE sellers ADD FOREIGN KEY(seller_zip_code_prefix) REFERENCES locations(geolocation_zip_code_prefix)""")
```

```python
cursor.execute("DELETE o FROM orders AS o LEFT JOIN customers AS c ON o.customer_id = c.customer_id WHERE c.customer_id IS NULL")
cursor.execute("""ALTER TABLE orders ADD FOREIGN KEY(customer_id) REFERENCES customers(customer_id)""")
```

```python
cursor.execute("DELETE p FROM products AS p LEFT JOIN translations AS t ON p.product_category_name = t.product_category_name" \
" WHERE t.product_category_name IS NULL")
cursor.execute("""ALTER TABLE products ADD FOREIGN KEY(product_category_name) REFERENCES translations(product_category_name)""")
```

```python
cursor.execute("DELETE i FROM order_items AS i LEFT JOIN orders AS o ON i.order_id = o.order_id WHERE o.order_id IS NULL")
cursor.execute("DELETE i FROM order_items AS i LEFT JOIN products AS p ON i.product_id = p.product_id WHERE p.product_id IS NULL")
cursor.execute("DELETE i FROM order_items AS i LEFT JOIN sellers AS s ON i.seller_id = s.seller_id WHERE s.seller_id IS NULL")
cursor.execute("""ALTER TABLE order_items ADD FOREIGN KEY(order_id) REFERENCES orders(order_id)""")
cursor.execute("""ALTER TABLE order_items ADD FOREIGN KEY(product_id) REFERENCES products(product_id)""")
cursor.execute("""ALTER TABLE order_items ADD FOREIGN KEY(seller_id) REFERENCES sellers(seller_id)""")
```
